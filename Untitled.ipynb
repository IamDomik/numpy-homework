```python
sizes = [10, 50, 100, 200, 500]
t_non_vec, t_vec = [], []

for size in sizes:
    matrix = np.random.randint(-5, 5, size=(size, size))
    # Замеряем невекторизованную
    t1 = timeit.timeit(lambda: functions.task1(matrix.tolist()), number=5)
    t_non_vec.append(t1 / 5)
    # Замеряем векторизованную
    t2 = timeit.timeit(lambda: functions_vectorized.task1(matrix), number=5)
    t_vec.append(t2 / 5)

plt.figure(figsize=(8, 5))
plt.plot(sizes, t_non_vec, label="Невекторизованная", marker='o')
plt.plot(sizes, t_vec, label="Векторизованная", marker='s')
plt.title("Задача 1: Произведение ненулевых на диагонали")
plt.xlabel("Размер матрицы")
plt.ylabel("Время (с)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
